```python
from app.features.contract_agent.schemas import AgentEvalInputCase, AgentEvalInputDataset, AgentEvalRunRequest
```

### Create the Evaluation Dataset

```python
with open("agent_eval_inputs.json", "r") as f:
    input_cases = [AgentEvalInputCase(**case) for case in json.load(f)]
    dataset = AgentEvalInputDataset(name="Agent Evaluation - Thomson Reuters EULA", cases=input_cases, metadata={"contract_filename": "thomson-reuters-eula.pdf"})
```

### Trigger the Evaluation Run via the Backend API

```python
request = AgentEvalRunRequest(eval_id=eval_id, run_name="Agent Evaluation - Thomson Reuters EULA", dataset=dataset)
```

```python
response = response = requests.post(url=f"{backend_url}/agent/evals", json=request.model_dump())
response.raise_for_status()
response.status_code
```
